```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib as mpl
import scienceplots
import matplotlib.ticker as mticker
```

```python

mpl.rcParams.update({
    "font.family": "Cambria",
    "font.size": 11,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "pdf.fonttype": 42,
    "ps.fonttype": 42
})

# --- Load data once ---
df1 = pd.read_csv("PPO.csv", header=None, names=["Step", "Progress(%)", "CashInHand"])
df2 = pd.read_csv("A2C.csv", header=None, names=["Step", "Progress(%)", "CashInHand"])
df3 = pd.read_csv("SAC.csv", header=None, names=["Step", "Progress(%)", "CashInHand"])

# --- Create side‑by‑side figure ---
fig, axs = plt.subplots(1, 2, figsize=(10.8, 4.0))

# =========================
# (a) Project Progress
# =========================
axs[0].plot(df1["Step"], df1["Progress(%)"], color="green", linewidth=1.8, label="PPO")
axs[0].plot(df2["Step"], df2["Progress(%)"], color="#3f79e5", linewidth=1.8, label="A2C", linestyle="--", dashes=(2,1))
axs[0].plot(df3["Step"], df3["Progress(%)"], color="#e93a3a", linewidth=1.8, label="SAC", linestyle="--", dashes=(7,3))

axs[0].set_xlabel("Day")
axs[0].set_ylabel("Project progress (%)")

axs[0].set_xlim(0, 35)
axs[0].set_ylim(-5, 105)

axs[0].grid(True, color='black', linestyle='--', linewidth=0.4, alpha=0.3)
axs[0].legend(loc="lower right")

# =========================
# (b) Cash Flow
# =========================
axs[1].plot(df1["Step"], df1["CashInHand"], color="green", linewidth=1.8, label="PPO")
axs[1].plot(df2["Step"], df2["CashInHand"], color="#3f79e5", linewidth=1.8, label="A2C", linestyle="--", dashes=(2,1))
axs[1].plot(df3["Step"], df3["CashInHand"], color="#e93a3a", linewidth=1.8, label="SAC", linestyle="--", dashes=(7,3))

axs[1].set_xlabel("Day")
axs[1].set_ylabel("Cash flow ($)")

axs[1].set_xlim(0, 35)
axs[1].set_ylim(-11000, 32000)

axs[1].axhline(y=0, color="#711f1f", linewidth=1.0)

axs[1].grid(True, color='black', linestyle='--', linewidth=0.4, alpha=0.3)
axs[1].legend(loc="lower right")


axs[0].text(0.5, -0.22, "(a)", transform=axs[0].transAxes,
            ha="center", va="center", fontsize=14, fontfamily="Times new roman")

axs[1].text(0.5, -0.22, "(b)", transform=axs[1].transAxes,
            ha="center", va="center", fontsize=14, fontfamily="Times new roman")


# --- Layout ---
plt.tight_layout()

# --- Save ---
plt.savefig("Fig9_combined_new3_Springer.png", dpi=600, bbox_inches='tight')

plt.show()
```
