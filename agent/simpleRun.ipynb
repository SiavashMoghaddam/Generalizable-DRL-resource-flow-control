```python
areaZone = 100
numZone = 4
numFloor = 5


for i in range(1):
    sim = AnyLogicSim("D:/PhD/Thesis_2/4_Paper/3_Experiments/Exp1/Exp1A/RL/Exported/model.jar",engine_overrides=dict(seed=1))
    status = sim.reset(cnfAreaZone=areaZone, cnfNumZone=numZone, cnfNumFloor=numFloor,
                       cnfPayPerUnit=450, cnfStorageCapacity=2, cnfRemoveFormC=1, cnfRemoveFormBF=14,
                       cnfCashStart=areaZone*numZone*450*1.2)
    done1 = False
    time = 0
    print(status.time)
    print(status.observation)
    while not done1:
        status = sim.take_action(actCapRebar=4, actCapForm=5, actCapConc=6, actMatRebar=7, actMatForm=8, actMatConc=9)
        done1 = sim.status().stop
        time += 1
        print(status.time)
        print(status.observation)
    print("Time: ",time, ",  Progress:",sim.observation()['obsProgress']*100,"%",",   Cash:", sim.observation()['obsCashInHand'])
sim._quit_app()



```
